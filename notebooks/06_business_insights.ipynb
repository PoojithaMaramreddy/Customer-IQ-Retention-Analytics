# Business Insights

## Objective

The objective of this analysis is to summarize customer behavior, purchasing patterns, sales performance, and churn-related observations from the previous analysis stages.

The insights are based on the cleaned transaction data, RFM segmentation, and churn analysis.

```python
import pandas as pd

df = pd.read_csv("../data/processed/online_retail_cleaned.csv")
rfm = pd.read_csv("../data/processed/customer_rfm.csv")
churn = pd.read_csv("../data/processed/churn_predictions.csv")

print("Cleaned data:", df.shape)
print("RFM data:", rfm.shape)
print("Churn predictions:", churn.shape)
```

```python
total_sales = (df["Quantity"] * df["Price"]).sum()

print("Total historical sales:", round(total_sales, 2))
```

## Overall Sales Performance

The cleaned transaction dataset contains total historical sales of approximately **17.37 million**.

This value is calculated as the sum of `Quantity × Price` across all cleaned transaction records.

This represents historical transaction sales value in the dataset and should not be interpreted as profit or future revenue.

## Customer Segment Distribution

The RFM analysis identified five customer segments:

- **Lost Customers:** 1,600 customers (27.22%)
- **Potential Loyalists:** 1,322 customers (22.49%)
- **Loyal Customers:** 1,278 customers (21.74%)
- **At Risk:** 1,209 customers (20.57%)
- **Champions:** 469 customers (7.98%)

Lost Customers represent the largest customer segment, while Champions represent the smallest segment.

## Monetary Contribution by Customer Segment

The distribution of historical monetary value varies substantially across customer segments.

- **Champions:** 46.88% of total historical monetary value
- **Loyal Customers:** 24.58%
- **Potential Loyalists:** 14.27%
- **At Risk:** 8.64%
- **Lost Customers:** 5.62%

Champions account for only **7.98% of customers** but contribute **46.88% of historical monetary value**.

This indicates that a relatively small group of customers generated a large share of the historical sales value in the dataset.

## Customer Behavior and Churn Observation

Customers who were classified as churned had different historical purchasing behavior compared with customers who made a purchase after the cutoff date.

Average values by churn status were:

- **Churned customers:** Recency = 273.72 days, Frequency = 3.05 invoices, Monetary = 1,134.62
- **Non-churned customers:** Recency = 119.56 days, Frequency = 9.25 invoices, Monetary = 4,598.24

In this dataset, churned customers generally had higher recency and lower purchase frequency and monetary value.

These are observed relationships in the dataset and should not be interpreted as proof of causation.

## High-Value Customer Observation

The top 10 customers by historical monetary value generated substantial transaction value and were primarily classified as Champions.

Among the top 10 customers:

- **8 customers** were classified as Champions.
- **2 customers** were classified as Loyal Customers.
- The highest-value customer generated **580,987.04** in historical monetary value across **145 invoices**.
- Another high-frequency customer recorded **398 invoices**.

This shows that the highest-value customers in the dataset generally had strong purchasing frequency and recent purchase activity.

## Monthly Sales Pattern

Monthly historical sales show noticeable variation across the observation period.

Key observations include:

- Sales activity was generally stronger during **September, October, and November**.
- **November 2010** recorded approximately **1.17 million** in historical sales value.
- **November 2011** recorded approximately **1.16 million**.
- October also showed relatively high sales in both years.
- Sales declined considerably during December in both years.

The monthly pattern suggests that sales activity was higher during the September-November period in this dataset.

## Average Customer Value by Segment

Average historical monetary value differs considerably across the RFM segments:

- **Champions:** 17,367.30
- **Loyal Customers:** 3,342.02
- **Potential Loyalists:** 1,876.07

Champions have the highest average historical monetary value, while Lost Customers have the lowest.

This difference helps demonstrate how customer segmentation can be used to understand variations in historical customer value.
