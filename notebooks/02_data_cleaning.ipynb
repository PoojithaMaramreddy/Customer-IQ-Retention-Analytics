# Data Cleaning

This notebook contains the data cleaning and preprocessing steps
performed on the Online Retail II dataset.

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/online_retail_II.csv")
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df=df.dropna(subset=['Customer ID'])
```

```python
df.shape
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df.dtypes
```

```python
(df["Quantity"] < 0).sum()
```

```python
df = df[df["Quantity"] > 0]
```

```python
df.shape
```

```python
(df["Price"] <= 0).sum()
```

```python
df = df[df["Price"] > 0]
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[["Quantity", "Price"]].describe()
```

## Cleaning Summary

The following data cleaning steps were performed:

- Removed duplicate records.
- Removed records with missing Customer ID.
- Converted InvoiceDate from string to datetime.
- Removed transactions with negative Quantity, which represent returns/cancellations for this analysis.
- Removed transactions with zero or negative Price.
- Verified that no missing values remain.
- Verified that no duplicate records remain.
- Performed final validation of Quantity and Price.

After cleaning, the dataset contains **779,425 rows and 8 columns**.

The cleaned dataset is now ready for Exploratory Data Analysis (EDA).
